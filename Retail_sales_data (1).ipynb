### Data Cleaning and Preprocessing

This cell checks for and removes any duplicate rows in the dataset to ensure data integrity, reporting the number of duplicates found.

```python
# Check and remove duplicates
duplicate_count = df.duplicated().sum()
print(f"Duplicates found: {duplicate_count}")

df = df.drop_duplicates()
```

This cell converts the 'Date' column to a datetime object and casts 'Customer ID', 'Gender', and 'Product Category' to appropriate categorical data types, preparing them for analysis.

```python
df['date'] = pd.to_datetime(df['date'])
df['customer_id'] = df['customer_id'].astype(str)
df['gender'] = df['gender'].astype('category')
df['product_category'] = df['product_category'].astype('category')
```

This cell standardizes the 'Gender' and 'Product Category' columns by stripping leading/trailing whitespace and converting their values to title case for consistency.

```python
df['gender'] = df['gender'].str.strip().str.title()
df['product_category'] = df['product_category'].str.strip().str.title()
```

This cell converts the 'Price per Unit' and 'Total Amount' columns to float data types to ensure accurate numerical operations.

```python
df['price_per_unit'] = df['price_per_unit'].astype(float)
df['total_amount'] = df['total_amount'].astype(float)
```

### Data Export

This markdown cell indicates that the data cleaning and preparation steps are complete, and the data is ready for further use, such as SQL operations.

The data is now cleaned and prepared for SQL. All column names are in snake_case, appropriate data types are assigned, and categorical values have been standardized.

This cell exports the cleaned and preprocessed DataFrame to a new CSV file named `retail_sales_cleaned.csv` without including the DataFrame index.

```python
output_filename = 'retail_sales_cleaned.csv'
df.to_csv(output_filename, index=False)
print(f"Data exported to {output_filename}")
```
